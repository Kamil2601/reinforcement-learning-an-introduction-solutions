### Exercise 11.1

Convert the equation of n-step off-policy TD (7.9) to semi-gradient form. Give accompanying definitions of the return for both the episodic and continuing cases.

**Answer**

* Update: $$\mathbf{w}_{t+n} = \mathbf{w}_{t+n-1} + \alpha \rho_{t:t+n-1} \left[G_{t:t+n} - \hat{v}(S_t, \mathbf{w}_{t+n-1}) \right] \nabla  \hat{v}(S_t, \mathbf{w}_{t+n-1})$$
* Episodic return: $$G_{t:t+n} = R_{t+1} + ... + \gamma^{n-1}R_{t+n} + \gamma^n \hat{v}(S_t, \mathbf{w}_{t+n-1})$$
* Continous return: $$G_{t:t+n} = R_{t+1} - \bar{R}_t + ... + R_{t+n}  - \bar{R}_{t+n-1} + \hat{v}(S_t, \mathbf{w}_{t+n-1})$$

### Exercise 11.2

Convert the equations of n-step $Q(\sigma)$ (7.11 and 7.17) to semi-gradient form. Give definitions that cover both the episodic and continuing cases.

**Answer**

Update: $$\mathbf{w}_{t+n} = \mathbf{w}_{t+n-1} + \alpha \rho_{t:t+n-1} \left[G_{t:t+n} - \hat{q}(S_t, A_t \mathbf{w}_{t+n-1}) \right] \nabla  \hat{q}(S_t, A_t \mathbf{w}_{t+n-1})$$

Episodic return: $$G_{t:t+n} = R_{t+1} + \gamma \left(\sigma_{t+1}\rho_{t+1} + (1 - \sigma_{t+1})\pi(A_{t+1}|S_{t+1})\right) \left(G_{t+1:h} -  \hat{q}(S_{t+1}, A_{t+1} \mathbf{w}_{h-1}\right) + \gamma \bar{V}_{h-1}(S_{t+1})$$
where: $$\bar{V}_{t}(s) = \sum_{a} \pi(a|s) \hat{q}(s, a, w_t)$$

$$G_{h:h} = \hat{q}(S_h, A_h, w_{h-1})$$

Continuous return:
$$
G_{t:t+n} = R_{t+1} - \bar{R}_t + \left(\sigma_{t+1}\rho_{t+1} + (1 - \sigma_{t+1})\pi(A_{t+1}|S_{t+1})\right) \left(G_{t+1:h} -  \hat{q}(S_{t+1}, A_{t+1} \mathbf{w}_{h-1}\right) + \bar{V}_{h-1}(S_{t+1})
$$
