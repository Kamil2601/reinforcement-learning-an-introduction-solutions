### Exercise 12.1

Just as the return can be written recursively in terms of the first reward and itself one-step later (3.9), so can the λ-return. Derive the analogous recursive relationship
from (12.2) and (12.1).

**Answer**

\begin{split}
G_t^{\lambda} &= (1 - \lambda)\sum_{n=1}^{\infty} \lambda G_{t:t+n} \\
&= (1 - \lambda) \left [G_{t:t+1} + \lambda G_{t:t+2} +  \lambda^2 G_{t:t+3} + ...\right] \\
&= (1 - \lambda) \left [
        (R_{t+1} + \gamma \hat{v}(S_{t+1}, \mathbf{w}_t)) 
        + \lambda (R_{t+1} + \gamma R_{t+2} + \gamma^2 \hat{v}(S_{t+2}, \mathbf{w}_{t+1}))
        + + \lambda^2 (R_{t+1} + \gamma R_{t+2} + \gamma^2 R_{t+3} + \gamma^3 \hat{v}(S_{t+2}, \mathbf{w}_{t+1})) + ...  \right] \\
&= (1-\lambda)[R_{t+1} + \lambda R_{t+2} + \lambda^2 R_{t+3} ...] + (1-\lambda)\gamma \hat{v}(S_{t+1}, \mathbf{w}_t)
    + (1 - \lambda)\gamma \lambda [G_{t+1:t+2} + G_{t+1:t+3} + ...] \\
&= R_{t+1} + (1-\lambda)\gamma \hat{v}(S_{t+1}, \mathbf{w}_t) + \gamma \lambda G_{t+1}^{\lambda}
\end{split}

### Exercise 12.2

The parameter $\lambda$ characterizes  how fast the exponential weighting in Figure 12.2 falls off, and thus how far into the future the $\lambda$-return algorithm looks in
determining its update. But a rate factor such as $\lambda$ is sometimes an awkward way of characterizing the speed of the decay. For some purposes it is better to specify a time
constant, or half-life. What is the equation relating $\lambda$ and the half-life, $\tau_{\lambda}$ , the time by which the weighting sequence will have fallen to half of its initial value?

**Answer**

We need to solve equation
$$
\lambda^{\tau_{\lambda}} = \frac{1}{2}
$$

$$
\ln(\lambda^{\tau_{\lambda}}) = \ln(\frac{1}{2})
$$

$$
\tau_{\lambda}\ln(\lambda) = -\ln(2)
$$

$$
\tau_{\lambda} = -\frac{\ln(2)}{\ln(\lambda)}
$$
